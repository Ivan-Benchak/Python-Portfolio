Flow rate of 1 L/min:

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Data1(4).csv")

df.columns = df.columns.str.replace('[^A-Za-z0-9]+', '', regex=True)
print(df.columns)

plt.figure(figsize=(10, 5))


plt.plot(df['TimeX'], df['Temperature1'], linestyle='-', linewidth=2, color='green', label='T_1(Inlet)')
plt.plot(df['TimeX'], df['Temperature2'], linestyle='-', linewidth=2, color='red',   label='T_2(Jacketed Vessel)')
plt.plot(df['TimeX'], df['Temperature3'], linestyle='-', linewidth=2, color='blue',  label='T_3(Outlet)')

plt.axhline(70, color='red', linestyle='--', label='Steady State at 70 °C') #Setted desired temperature on the Apparatus for the Jacketed Vessel (Refer only to the T_2 curve).

plt.xlabel("Time / seconds(s)")
plt.ylabel("Temperature / °C")
plt.title("Temperature vs Time for the Flow Rate of 1L/min")
plt.legend()
plt.grid(True)

plt.xlim(0, 1900)
plt.ylim(18, 75)
plt.savefig("Exp_3_1_L_min_-1.png", dpi=150, bbox_inches="tight")
plt.show()
```

Log - Log Relashinship for 1L/min

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv("Data3.csv")
df.columns = df.columns.str.replace('[^A-Za-z0-9]+', '', regex=True)


coefficients = np.polyfit(df['TimeX'], df['Y'], 1)
p = np.poly1d(coefficients)
slope = coefficients[0]
intercept = coefficients[1]


plt.figure(figsize=(10, 5))

plt.scatter(df['TimeX'], df['Y'], s=10, color='green', label="Data Points")
plt.plot(df['TimeX'], df['Y'], linewidth = 1, linestyle = "--")


plt.plot(df['TimeX'], p(df['TimeX']), linestyle='-', color='orange',
         label=f"Linear Fit: $Y = {slope:.4f}t + {intercept:.2f}$", linewidth=2)

plt.xlabel("Time / seconds (s)")
plt.ylabel(r'$\ln\!\left(\frac{T_{h_m} - T(t)}{T_{h_m} - T_0}\right) = \frac{t}{\tau}$', fontsize=14)

plt.title("Log-Log Plot of Equation (9) for the Flow Rate of 1L/min")
plt.legend()
plt.grid(True)


plt.xlim(0, 1900)
plt.ylim(-8, 0.5)

plt.savefig("Exp_3_1_L_min_-3(1)_with_fit.png", dpi=150, bbox_inches="tight")
plt.show()
```
