```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter


df = pd.read_csv("Data6.csv")


df.columns = df.columns.str.replace('[^A-Za-z0-9]+', '', regex=True)


df['TemperatureY_smooth'] = savgol_filter(df['TemperatureY'], window_length=11, polyorder=3)

x = df['TimeX']  
y = df['TemperatureY_smooth'] 

plt.figure(figsize=(15, 4))

plt.plot(x, y, linestyle='-', linewidth=2, color='blue', label='Smoothed Temperature vs Time')

plt.scatter(df['TimeX'], df['TemperatureY'], color='red', label='Original Data Points', s=15)

plt.axhline(47, color='green', linestyle='--', label='Steady State at 47 °C') 
plt.xlabel("Time / seconds(s)")
plt.ylabel("Temperature / °C")
plt.title("Temperature vs Time Graph")
plt.legend()
plt.grid(True)

plt.savefig("Exp_5.png", dpi=150, bbox_inches="tight")
plt.show()
```
